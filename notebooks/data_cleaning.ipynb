TotalCharges are in string format, we've to change it to numeric

```python
df.nunique()
```

```python
print(df['PaymentMethod'].value_counts())
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
```

```python
numeric_col = ["tenure", "MonthlyCharges", "TotalCharges"]

categorical_col = [col for col in df.columns if col not in numeric_col + ["Churn"]]
print("Numeric columns:", numeric_col)
print("Categorical columns:", categorical_col)
```

```python
df = df.drop(columns=['customerID'])
```

customerID is just a unique ID, it has no pattern to learn from.

```python
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

print("Churn value counts:")
print(df['Churn'].value_counts())
```

This means the Churn column values don't exactly match 'Yes' and 'No'. There might be extra spaces or different casing.

```python
# Save the cleaned data to processed folder so we don't redo this every time
df.to_csv("/Users/syedalihussain/Documents/customer_churn/data/processed/cleaned_churn.csv", index=False)

print("Cleaned data saved!")
```
