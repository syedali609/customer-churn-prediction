```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("/Users/syedalihussain/Documents/customer_churn/data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Shape:", df.shape)
df.head()
```

```python
# df = df.drop(columns=['customerID'])
# TotalCharges is stored as text, convert it to number
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
median_val = df['TotalCharges'].median()
df['TotalCharges'] = df['TotalCharges'].fillna(median_val)
```

```python
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
print(df['Churn'].value_counts())
```

```python
text_cols = df.select_dtypes(include='str').columns.tolist()
print("Text columns remaining:", text_cols)
```

```python
df.info()
```

```python
le = LabelEncoder()
for col in text_cols:
    df[col] = le.fit_transform(df[col])
    print(f'{col} encoded')

print(df.dtypes)
```

```python
X = df.drop(columns=['Churn'])
y = df['Churn']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
scaler = StandardScaler()
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
X[num_cols] = scaler.fit_transform(X[num_cols])
X.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size= 0.2, stratify=y
)
print("Train size:", X_train.shape)
print("Test size:", X_test.shape)
```

```python
X_train = pd.get_dummies(X_train)
X_test = pd.get_dummies(X_test)

X_train, X_test = X_train.align(X_test, join='left', axis=1, fill_value=0)
```

```python
train_data = X_train.copy()
train_data['Churn'] = y_train.values

test_data = y_train.copy()
test_data['Churn'] = y_test.values

train_data.to_csv("/Users/syedalihussain/Documents/customer_churn/data/processed/train.csv", index=False)
test_data.to_csv("/Users/syedalihussain/Documents/customer_churn/data/processed/test.csv", index=False)

print("Train and test data saved to data/processed/!")
```
